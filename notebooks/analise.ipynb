```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parents[0] if Path.cwd().name == "notebooks" else Path.cwd()))

import pandas as pd
import matplotlib.pyplot as plt
from src.config import GOLD_CSV_DIR

overview = pd.read_csv(GOLD_CSV_DIR / "overview.csv")
market = pd.read_csv(GOLD_CSV_DIR / "market_structure.csv")
salary = pd.read_csv(GOLD_CSV_DIR / "salary_role_seniority.csv")
gender = pd.read_csv(GOLD_CSV_DIR / "gender_gap.csv")
tech = pd.read_csv(GOLD_CSV_DIR / "tech_adoption.csv")
ai = pd.read_csv(GOLD_CSV_DIR / "ai_adoption.csv")
region = pd.read_csv(GOLD_CSV_DIR / "region_work_model.csv")
sector = pd.read_csv(GOLD_CSV_DIR / "sector.csv")
display(overview)
```

```python
ano = 2025
m = market[market["survey_year"] == ano].sort_values("professionals")
plt.barh(m["job_family"], m["share_pct"])
plt.title("mercado 2025")
plt.xlabel("%")
plt.show()

s = salary[(salary["survey_year"] == ano) & (salary["seniority_group"] == "Sênior")].sort_values("avg_salary")
plt.barh(s["job_family"], s["avg_salary"])
plt.title("salario senior")
plt.show()

g = gender[gender["survey_year"] == ano]
print(g)

t = tech[tech["survey_year"] == ano].sort_values("adoption_pct")
plt.barh(t["technology"], t["adoption_pct"])
plt.title("ferramentas 2025")
plt.show()
```
